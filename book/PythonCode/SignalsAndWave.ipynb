```python
cos_sig = thinkdsp.CosSignal(freq=440, amp=1.0, offset=0)
sin_sig = thinkdsp.SinSignal(freq=880, amp=0.5, offset=0)
mix = sin_sig + cos_sig
wave = mix.make_wave(duration=0.5, start=0, framerate=11025)
```

```python
wave.plot()
plt.show()
```

```python
period = mix.period
print(period)
```

```python
segment = wave.segment(start=0, duration=period*3)
```

```python
segment.plot()
t_start = segment.start
t_end = segment.end

# Generate cycle start times
cycle_starts = np.arange(t_start, t_end, period)

# Add vertical lines at cycle starts
for t in cycle_starts:
    plt.axvline(x=t, color="red", linestyle="--")

# Optionally mark cycle ends (start + period)
for t in cycle_starts:
    plt.axvline(x=t+period, color="green", linestyle=":")

plt.show()
```
